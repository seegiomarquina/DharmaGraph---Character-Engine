```python


# ------------------------- 1. INSTALL ----------------------------

import sys
import subprocess

packages = [
    "pandas",
    "numpy",
    "networkx",
    "plotly",
    "scikit-learn",
    "matplotlib",
    "seaborn",
    "tqdm"
]

subprocess.check_call(
    [sys.executable, "-m", "pip", "install", "-q"] + packages
)

# ------------------------- 2. IMPORTS ----------------------------

import os
import json
import re
import sqlite3
import requests
import warnings
import numpy as np
import pandas as pd
import networkx as nx
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from IPython.display import display, HTML

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)

PROJECT = Path("DharmaGraph")
DATA_DIR = PROJECT / "data"
OUTPUT_DIR = PROJECT / "outputs"
PLOT_DIR = OUTPUT_DIR / "plots"
SQL_DIR = PROJECT / "sql"

for folder in [DATA_DIR, OUTPUT_DIR, PLOT_DIR, SQL_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("DHARMAGRAPH INITIALIZED")

# ---------------------- 3. DOWNLOAD DATA -------------------------

DATA_URLS = [
    "https://raw.githubusercontent.com/Ashutosh-Vijay/Valmiki_Ramayan_Dataset/main/data/Valmiki_Ramayan_Shlokas.json",
    "https://raw.githubusercontent.com/AshuVj/Valmiki_Ramayan_Dataset/main/data/Valmiki_Ramayan_Shlokas.json"
]

DATA_FILE = DATA_DIR / "ramayana_verses.json"

if not DATA_FILE.exists():
    downloaded = False

    for url in DATA_URLS:
        try:
            response = requests.get(url, timeout=120)
            response.raise_for_status()

            data_test = response.json()

            if isinstance(data_test, list) and len(data_test) > 0:
                with open(DATA_FILE, "w", encoding="utf-8") as f:
                    json.dump(data_test, f, ensure_ascii=False)

                downloaded = True
                print("Dataset downloaded successfully:", url)
                break

        except Exception as e:
            print("Download attempt failed:", str(e)[:150])

    if not downloaded:
        raise RuntimeError(
            "Dataset download failed. Open the official repository "
            "and verify the current JSON file path."
        )
else:
    print("Existing dataset found.")

with open(DATA_FILE, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

if isinstance(raw_data, dict):
    for key in ["data", "verses", "shlokas", "records"]:
        if key in raw_data and isinstance(raw_data[key], list):
            raw_data = raw_data[key]
            break

if not isinstance(raw_data, list):
    raise ValueError("Unexpected dataset structure. Inspect the JSON.")

print("Raw records:", len(raw_data))

# ---------------------- 4. DATA CLEANING --------------------------

df = pd.DataFrame(raw_data)

required_columns = [
    "kanda", "sarga", "shloka",
    "shloka_text", "transliteration",
    "translation", "explanation"
]

for col in required_columns:
    if col not in df.columns:
        df[col] = ""

for col in required_columns:
    df[col] = df[col].fillna("").astype(str)

df["sarga"] = pd.to_numeric(df["sarga"], errors="coerce")
df["shloka"] = pd.to_numeric(df["shloka"], errors="coerce")

df["search_text"] = (
    df["translation"] + " " + df["explanation"]
).str.strip()

df["reference"] = (
    df["kanda"] + " | Sarga " +
    df["sarga"].fillna(0).astype(int).astype(str) +
    " | Shloka " +
    df["shloka"].fillna(0).astype(int).astype(str)
)

df = df.drop_duplicates(
    subset=["kanda", "sarga", "shloka"],
    keep="first"
).reset_index(drop=True)

df = df[
    (df["kanda"].str.len() > 0) |
    (df["shloka_text"].str.len() > 0)
].copy()

df.to_csv(DATA_DIR / "cleaned_verses.csv", index=False)

print("\nCLEANED DATASET")
print("Rows:", len(df))
print("Columns:", df.shape[1])

display(df.head(5))

print("\nKANDA DISTRIBUTION")
display(df["kanda"].value_counts().to_frame("verse_records"))

# ---------------------- 5. EDA -----------------------------------

plt.figure(figsize=(12, 5))
df["kanda"].value_counts().plot(kind="bar")
plt.title("Ramayana Verse Records by Kanda")
plt.xlabel("Kanda")
plt.ylabel("Number of Records")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.savefig(PLOT_DIR / "kanda_distribution.png", dpi=150)
plt.show()

df["explanation_length"] = df["explanation"].str.split().str.len()

plt.figure(figsize=(10, 5))
sns.histplot(
    df.loc[df["explanation_length"] > 0, "explanation_length"],
    bins=40,
    kde=True
)
plt.title("Distribution of Explanation Lengths")
plt.xlabel("Number of Words")
plt.tight_layout()
plt.savefig(PLOT_DIR / "explanation_lengths.png", dpi=150)
plt.show()

# ---------------------- 6. KNOWLEDGE GRAPH DATA ------------------
# This is a manually curated starter ontology.
# Relationships are research annotations, not automatically
# extracted or independently verified facts.
#
# Add source references to the annotation sheet before publication.

characters = {
    "Rama": "Character",
    "Sita": "Character",
    "Lakshmana": "Character",
    "Hanuman": "Character",
    "Ravana": "Character",
    "Bharata": "Character",
    "Shatrughna": "Character",
    "Dasharatha": "Character",
    "Kaushalya": "Character",
    "Kaikeyi": "Character",
    "Sumitra": "Character",
    "Jatayu": "Character",
    "Sugriva": "Character",
    "Vali": "Character",
    "Vibhishana": "Character",
    "Angada": "Character",
    "Jambavan": "Character",
    "Indrajit": "Character",
    "Kumbhakarna": "Character",
    "Janaka": "Character",
    "Vishvamitra": "Character",
    "Narada": "Character",
    "Valmiki": "Character",
    "Shabari": "Character",
    "Surpanakha": "Character"
}

locations = {
    "Ayodhya": "Location",
    "Mithila": "Location",
    "Dandaka Forest": "Location",
    "Panchavati": "Location",
    "Kishkindha": "Location",
    "Rishyamukha": "Location",
    "Lanka": "Location",
    "Ashoka Vatika": "Location",
    "Chitrakuta": "Location",
    "Rama Setu": "Location",
    "Valmiki Ashram": "Location"
}

events = {
    "Rama's Birth": "Event",
    "Sita's Swayamvara": "Event",
    "Rama's Exile": "Event",
    "Forest Journey": "Event",
    "Sita's Abduction": "Event",
    "Jatayu's Encounter": "Event",
    "Meeting with Hanuman": "Event",
    "Alliance with Sugriva": "Event",
    "Search for Sita": "Event",
    "Hanuman's Lanka Mission": "Event",
    "Construction of Rama Setu": "Event",
    "Battle of Lanka": "Event",
    "Ravana's Defeat": "Event",
    "Rama's Return to Ayodhya": "Event"
}

# Relationship format:
# source, target, relationship, evidence_note
# These are curated starting annotations.

relationship_data = [
    ("Rama", "Sita", "spouse", "Traditional Ramayana narrative"),
    ("Rama", "Lakshmana", "brother", "Traditional Ramayana narrative"),
    ("Rama", "Bharata", "brother", "Traditional Ramayana narrative"),
    ("Rama", "Shatrughna", "brother", "Traditional Ramayana narrative"),
    ("Dasharatha", "Rama", "father_of", "Traditional Ramayana narrative"),
    ("Dasharatha", "Bharata", "father_of", "Traditional Ramayana narrative"),
    ("Dasharatha", "Lakshmana", "father_of", "Traditional Ramayana narrative"),
    ("Dasharatha", "Shatrughna", "father_of", "Traditional Ramayana narrative"),
    ("Kaushalya", "Rama", "mother_of", "Traditional Ramayana narrative"),
    ("Kaikeyi", "Bharata", "mother_of", "Traditional Ramayana narrative"),
    ("Sumitra", "Lakshmana", "mother_of", "Traditional Ramayana narrative"),
    ("Sumitra", "Shatrughna", "mother_of", "Traditional Ramayana narrative"),
    ("Janaka", "Sita", "father_of", "Traditional Ramayana narrative"),
    ("Ravana", "Vibhishana", "brother", "Traditional Ramayana narrative"),
    ("Ravana", "Kumbhakarna", "brother", "Traditional Ramayana narrative"),
    ("Ravana", "Surpanakha", "sibling", "Traditional Ramayana narrative"),
    ("Rama", "Hanuman", "ally", "Traditional Ramayana narrative"),
    ("Rama", "Sugriva", "ally", "Traditional Ramayana narrative"),
    ("Rama", "Jatayu", "ally", "Traditional Ramayana narrative"),
    ("Rama", "Shabari", "meets", "Traditional Ramayana narrative"),
    ("Hanuman", "Sugriva", "ally", "Traditional Ramayana narrative"),
    ("Sugriva", "Vali", "brother", "Traditional Ramayana narrative"),
    ("Vibhishana", "Rama", "ally", "Traditional Ramayana narrative"),
    ("Ravana", "Rama", "opponent", "Traditional Ramayana narrative"),
    ("Ravana", "Sita", "abductor_of", "Traditional Ramayana narrative"),
    ("Lakshmana", "Indrajit", "opponent", "Traditional Ramayana narrative"),
    ("Rama", "Vishvamitra", "teacher_and_guide", "Traditional Ramayana narrative"),
    ("Valmiki", "Rama", "narrates_story_of", "Traditional Ramayana tradition"),
    ("Narada", "Valmiki", "guides", "Traditional Ramayana narrative")
]

event_participants = {
    "Rama's Birth": ["Rama", "Dasharatha", "Kaushalya"],
    "Sita's Swayamvara": ["Rama", "Sita", "Janaka"],
    "Rama's Exile": ["Rama", "Sita", "Lakshmana", "Dasharatha", "Kaikeyi", "Bharata"],
    "Forest Journey": ["Rama", "Sita", "Lakshmana"],
    "Sita's Abduction": ["Ravana", "Sita", "Jatayu", "Rama", "Lakshmana"],
    "Jatayu's Encounter": ["Jatayu", "Ravana", "Sita"],
    "Meeting with Hanuman": ["Rama", "Lakshmana", "Hanuman", "Sugriva"],
    "Alliance with Sugriva": ["Rama", "Sugriva", "Hanuman", "Lakshmana"],
    "Search for Sita": ["Hanuman", "Sugriva", "Angada", "Jambavan"],
    "Hanuman's Lanka Mission": ["Hanuman", "Sita", "Ravana"],
    "Construction of Rama Setu": ["Rama", "Lakshmana", "Hanuman", "Sugriva"],
    "Battle of Lanka": ["Rama", "Lakshmana", "Hanuman", "Ravana", "Vibhishana", "Indrajit", "Kumbhakarna"],
    "Ravana's Defeat": ["Rama", "Ravana", "Vibhishana"],
    "Rama's Return to Ayodhya": ["Rama", "Sita", "Lakshmana", "Bharata", "Hanuman"]
}

event_locations = {
    "Rama's Birth": ["Ayodhya"],
    "Sita's Swayamvara": ["Mithila"],
    "Rama's Exile": ["Ayodhya", "Chitrakuta"],
    "Forest Journey": ["Dandaka Forest", "Panchavati"],
    "Sita's Abduction": ["Panchavati", "Lanka"],
    "Jatayu's Encounter": ["Dandaka Forest"],
    "Meeting with Hanuman": ["Rishyamukha"],
    "Alliance with Sugriva": ["Kishkindha"],
    "Search for Sita": ["Kishkindha"],
    "Hanuman's Lanka Mission": ["Lanka", "Ashoka Vatika"],
    "Construction of Rama Setu": ["Rama Setu"],
    "Battle of Lanka": ["Lanka"],
    "Ravana's Defeat": ["Lanka"],
    "Rama's Return to Ayodhya": ["Ayodhya"]
}

event_kandas = {
    "Rama's Birth": "Bala Kanda",
    "Sita's Swayamvara": "Bala Kanda",
    "Rama's Exile": "Ayodhya Kanda",
    "Forest Journey": "Aranya Kanda",
    "Sita's Abduction": "Aranya Kanda",
    "Jatayu's Encounter": "Aranya Kanda",
    "Meeting with Hanuman": "Kishkindha Kanda",
    "Alliance with Sugriva": "Kishkindha Kanda",
    "Search for Sita": "Kishkindha Kanda",
    "Hanuman's Lanka Mission": "Sundara Kanda",
    "Construction of Rama Setu": "Yuddha Kanda",
    "Battle of Lanka": "Yuddha Kanda",
    "Ravana's Defeat": "Yuddha Kanda",
    "Rama's Return to Ayodhya": "Yuddha Kanda"
}

# ---------------------- 7. BUILD GRAPH ---------------------------

G = nx.MultiDiGraph()

for name, kind in characters.items():
    G.add_node(name, node_type=kind)

for name, kind in locations.items():
    G.add_node(name, node_type=kind)

for name, kind in events.items():
    G.add_node(name, node_type=kind)

for source, target, relation, note in relationship_data:
    G.add_edge(
        source,
        target,
        relation=relation,
        evidence_note=note,
        annotation_type="curated"
    )

for event, people in event_participants.items():
    for person in people:
        G.add_edge(
            person,
            event,
            relation="participates_in",
            annotation_type="curated"
        )

for event, places in event_locations.items():
    for place in places:
        G.add_edge(
            event,
            place,
            relation="occurs_at",
            annotation_type="curated"
        )

for event, kanda in event_kandas.items():
    G.add_edge(
        event,
        kanda,
        relation="associated_with_kanda",
        annotation_type="curated"
    )

print("\nGRAPH CREATED")
print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())

# ---------------------- 8. GRAPH TABLES --------------------------

node_rows = []

for node, attributes in G.nodes(data=True):
    node_rows.append({
        "node": node,
        "node_type": attributes.get("node_type", "Kanda")
    })

edge_rows = []

for source, target, key, attributes in G.edges(
    keys=True, data=True
):
    edge_rows.append({
        "source": source,
        "target": target,
        "relationship": attributes.get("relation", "related_to"),
        "annotation_type": attributes.get("annotation_type", "curated"),
        "evidence_note": attributes.get("evidence_note", "")
    })

nodes_df = pd.DataFrame(node_rows)
edges_df = pd.DataFrame(edge_rows)

nodes_df.to_csv(OUTPUT_DIR / "graph_nodes.csv", index=False)
edges_df.to_csv(OUTPUT_DIR / "graph_edges.csv", index=False)

# ---------------------- 9. GRAPH ANALYTICS -----------------------

# Convert to simple undirected graph for centrality analysis.
UG = nx.Graph()

for source, target, data in G.edges(data=True):
    UG.add_edge(source, target)

degree = nx.degree_centrality(UG)
betweenness = nx.betweenness_centrality(UG)
closeness = nx.closeness_centrality(UG)

centrality_df = pd.DataFrame({
    "character_or_entity": list(UG.nodes()),
    "degree_centrality": [degree[n] for n in UG.nodes()],
    "betweenness_centrality": [betweenness[n] for n in UG.nodes()],
    "closeness_centrality": [closeness[n] for n in UG.nodes()],
    "node_type": [
        G.nodes[n].get("node_type", "Kanda")
        for n in UG.nodes()
    ]
})

centrality_df = centrality_df.sort_values(
    "degree_centrality", ascending=False
)

centrality_df.to_csv(
    OUTPUT_DIR / "centrality_metrics.csv", index=False
)

print("\nTOP CONNECTED ENTITIES")
display(centrality_df.head(15))

# ---------------------- 10. CHARACTER EXPLORER -------------------

def explore_character(name):
    matches = [
        n for n in G.nodes()
        if n.lower() == name.lower()
    ]

    if not matches:
        print("Entity not found:", name)
        print("Try:", ", ".join(characters.keys()))
        return

    entity = matches[0]

    print("\n" + "=" * 55)
    print("DHARMAGRAPH ENTITY EXPLORER")
    print("=" * 55)
    print("Entity:", entity)
    print("Type:", G.nodes[entity].get("node_type", "Unknown"))

    print("\nDIRECT CONNECTIONS")

    for source, target, data in G.in_edges(entity, data=True):
        print(
            f"{source} --[{data.get('relation')}]--> {entity}"
        )

    for source, target, data in G.out_edges(entity, data=True):
        print(
            f"{entity} --[{data.get('relation')}]--> {target}"
        )

    print("\nCENTRALITY")
    row = centrality_df[
        centrality_df["character_or_entity"] == entity
    ]

    if not row.empty:
        display(row)

# ---------------------- 11. NLP VERSE SEARCH ---------------------

# TF-IDF search across English translations and explanations.
# This is lexical retrieval, not generative AI.

search_corpus = (
    df["search_text"].fillna("").astype(str).tolist()
)

valid_indices = [
    i for i, text in enumerate(search_corpus)
    if len(text.strip()) > 0
]

if len(valid_indices) == 0:
    raise ValueError("No usable English text for NLP search.")

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=30000,
    ngram_range=(1, 2),
    min_df=2
)

tfidf_matrix = vectorizer.fit_transform(
    [search_corpus[i] for i in valid_indices]
)

def search_ramayana(query, top_k=5):
    query_vector = vectorizer.transform([query])

    similarities = cosine_similarity(
        query_vector, tfidf_matrix
    ).flatten()

    top_positions = np.argsort(similarities)[::-1][:top_k]

    results = []

    for pos in top_positions:
        original_idx = valid_indices[pos]
        row = df.iloc[original_idx]

        results.append({
            "similarity": float(similarities[pos]),
            "reference": row["reference"],
            "kanda": row["kanda"],
            "shloka_text": row["shloka_text"],
            "translation": row["translation"],
            "explanation": row["explanation"]
        })

    result_df = pd.DataFrame(results)
    return result_df

# ---------------------- 12. SQL DATABASE -------------------------

DB_FILE = OUTPUT_DIR / "dharmagraph.db"

conn = sqlite3.connect(DB_FILE)

nodes_df.to_sql("nodes", conn, if_exists="replace", index=False)
edges_df.to_sql("relationships", conn, if_exists="replace", index=False)

df[[
    "kanda", "sarga", "shloka",
    "shloka_text", "translation",
    "explanation", "reference"
]].to_sql("verses", conn, if_exists="replace", index=False)

SQL_QUERIES = {
    "entity_counts": """
        SELECT node_type, COUNT(*) AS total_entities
        FROM nodes
        GROUP BY node_type
        ORDER BY total_entities DESC;
    """,

    "relationship_counts": """
        SELECT relationship, COUNT(*) AS total
        FROM relationships
        GROUP BY relationship
        ORDER BY total DESC;
    """,

    "most_connected_entities": """
        SELECT source AS entity, COUNT(*) AS connections
        FROM relationships
        GROUP BY source
        ORDER BY connections DESC
        LIMIT 10;
    """,

    "verse_counts": """
        SELECT kanda, COUNT(*) AS total_verses
        FROM verses
        GROUP BY kanda
        ORDER BY total_verses DESC;
    """
}

for query_name, query in SQL_QUERIES.items():
    print("\nSQL REPORT:", query_name)
    display(pd.read_sql_query(query, conn))

with open(SQL_DIR / "analysis_queries.sql", "w") as f:
    for name, query in SQL_QUERIES.items():
        f.write(f"-- {name}\n{query}\n\n")

# ---------------------- 13. INTERACTIVE GRAPH --------------------

# Spring layout creates a visual network.
# It does not represent geographical positions or story chronology.

pos = nx.spring_layout(
    UG,
    seed=SEED,
    k=0.85,
    iterations=100
)

edge_x = []
edge_y = []

for source, target in UG.edges():
    x0, y0 = pos[source]
    x1, y1 = pos[target]

    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

edge_trace = go.Scatter(
    x=edge_x,
    y=edge_y,
    mode="lines",
    line=dict(width=0.7, color="rgba(140,150,170,0.45)"),
    hoverinfo="none"
)

type_colors = {
    "Character": "#3B82F6",
    "Location": "#10B981",
    "Event": "#F59E0B",
    "Kanda": "#A855F7"
}

node_x = []
node_y = []
node_colors = []
node_sizes = []
node_text = []

for node in UG.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)

    node_type = G.nodes[node].get("node_type", "Kanda")
    node_colors.append(type_colors.get(node_type, "#94A3B8"))

    node_sizes.append(
        12 + 30 * degree.get(node, 0)
    )

    node_text.append(
        f"{node}<br>Type: {node_type}"
        f"<br>Degree centrality: {degree.get(node, 0):.3f}"
    )

node_trace = go.Scatter(
    x=node_x,
    y=node_y,
    mode="markers+text",
    text=list(UG.nodes()),
    textposition="top center",
    textfont=dict(size=9),
    hovertext=node_text,
    hoverinfo="text",
    marker=dict(
        size=node_sizes,
        color=node_colors,
        line=dict(width=1, color="white")
    )
)

fig = go.Figure(
    data=[edge_trace, node_trace],
    layout=go.Layout(
        title="DHARMAGRAPH — Ramayana Knowledge Network",
        template="plotly_dark",
        showlegend=False,
        hovermode="closest",
        margin=dict(b=20, l=20, r=20, t=60),
        xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        height=850
    )
)

fig.show()

fig.write_html(
    str(OUTPUT_DIR / "dharmagraph_network.html")
)

# ---------------------- 14. CHARACTER TYPE ANALYSIS -------------

plt.figure(figsize=(9, 5))

sns.countplot(
    data=nodes_df,
    x="node_type",
    order=nodes_df["node_type"].value_counts().index
)

plt.title("Knowledge Graph Entity Distribution")
plt.xlabel("Entity Type")
plt.ylabel("Number of Entities")
plt.tight_layout()

plt.savefig(PLOT_DIR / "entity_distribution.png", dpi=150)
plt.show()

# ---------------------- 15. COMMUNITY DETECTION -----------------

communities = list(
    nx.community.greedy_modularity_communities(UG)
)

community_rows = []

for community_id, members in enumerate(communities, start=1):
    for member in members:
        community_rows.append({
            "community": community_id,
            "entity": member,
            "entity_type": G.nodes[member].get(
                "node_type", "Kanda"
            )
        })

community_df = pd.DataFrame(community_rows)

community_df.to_csv(
    OUTPUT_DIR / "graph_communities.csv", index=False
)

print("\nGRAPH COMMUNITIES")
display(community_df.head(30))

# ---------------------- 16. EVENT TIMELINE -----------------------

timeline_df = pd.DataFrame([
    {
        "event": event,
        "kanda": event_kandas.get(event, "Unknown"),
        "participants": ", ".join(event_participants.get(event, [])),
        "locations": ", ".join(event_locations.get(event, []))
    }
    for event in events
])

timeline_df.to_csv(
    OUTPUT_DIR / "event_timeline.csv", index=False
)

display(timeline_df)

# ---------------------- 17. EXPORT FINAL DATA --------------------

centrality_df.to_csv(
    OUTPUT_DIR / "centrality_metrics.csv", index=False
)

nodes_df.to_csv(
    OUTPUT_DIR / "graph_nodes.csv", index=False
)

edges_df.to_csv(
    OUTPUT_DIR / "graph_edges.csv", index=False
)

timeline_df.to_csv(
    OUTPUT_DIR / "event_timeline.csv", index=False
)

# Save graph in reusable JSON format.
graph_json = nx.node_link_data(G)

with open(
    OUTPUT_DIR / "ramayana_knowledge_graph.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(graph_json, f, ensure_ascii=False, indent=2)

# Save graph in GraphML format.
nx.write_graphml(
    G,
    OUTPUT_DIR / "ramayana_knowledge_graph.graphml"
)

# Save metadata.
metadata = {
    "project": "DharmaGraph",
    "dataset": "Valmiki Ramayana structured JSON",
    "dataset_source": "https://github.com/Ashutosh-Vijay/Valmiki_Ramayan_Dataset",
    "verse_records_after_cleaning": int(len(df)),
    "graph_nodes": int(G.number_of_nodes()),
    "graph_edges": int(G.number_of_edges()),
    "centrality_method": "NetworkX degree, betweenness, closeness",
    "verse_search": "TF-IDF cosine similarity",
    "relationship_method": "Manually curated starter annotations",
    "random_seed": SEED
}

with open(OUTPUT_DIR / "project_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

conn.close()

# ---------------------- 18. INTERACTIVE COMMANDS -----------------

print("\n" + "=" * 65)
print("DHARMAGRAPH READY")
print("=" * 65)

print("""
TRY THESE COMMANDS:

1. Explore a character:
   explore_character("Hanuman")

2. Search Ramayana verses:
   search_ramayana("Hanuman travels to Lanka", top_k=5)

3. Explore another character:
   explore_character("Ravana")

4. View centrality rankings:
   display(centrality_df.head(15))

5. View all graph relationships:
   display(edges_df)

6. View event timeline:
   display(timeline_df)

7. View community assignments:
   display(community_df)

OUTPUT LOCATION:
DharmaGraph/outputs/

IMPORTANT:
The relationship graph is a curated starter ontology.
Verify and expand annotations with reliable textual references.
TF-IDF search is retrieval, not an AI-generated answer.
""")

print("PROJECT EXECUTION FINISHED")
print("Project folder:", PROJECT.resolve())

# ---------------------- 19. ZIP EXPORT ---------------------------

import shutil

zip_path = shutil.make_archive(
    "DharmaGraph_GitHub_Outputs",
    "zip",
    root_dir=".",
    base_dir="DharmaGraph"
)

print("\nZIP CREATED:", zip_path)
print("Download it from the Colab Files panel.")
```

```python

# ============================================================
# DHARMAGRAPH 2.0 — ADVANCED SEMANTIC SEARCH & RAMAYANA Q&A
# ============================================================

!pip -q install sentence-transformers

import numpy as np
import pandas as pd
import torch
from sentence_transformers import SentenceTransformer
from IPython.display import display, Markdown

# 1. LOAD SEMANTIC EMBEDDING MODEL
print("Loading semantic intelligence model...")

semantic_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# 2. PREPARE VERSE DOCUMENTS
semantic_df = df.copy().reset_index(drop=True)

def get_verse_text(row):
    fields = [
        row.get("reference", ""),
        row.get("translation", ""),
        row.get("explanation", "")
    ]
    return " ".join(
        str(value) for value in fields
        if pd.notna(value) and str(value).strip()
    )

semantic_df["semantic_text"] = semantic_df.apply(
    get_verse_text, axis=1
)

semantic_df = semantic_df[
    semantic_df["semantic_text"].str.strip().ne("")
].reset_index(drop=True)

# 3. GENERATE SEMANTIC EMBEDDINGS
print("Generating verse embeddings...")

verse_embeddings = semantic_model.encode(
    semantic_df["semantic_text"].tolist(),
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

print("Total indexed verses:", len(semantic_df))
print("Embedding dimensions:", verse_embeddings.shape)

# 4. SEMANTIC SEARCH ENGINE
def semantic_search(query, top_k=5):
    query_embedding = semantic_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores = np.dot(
        verse_embeddings,
        query_embedding[0]
    )

    best_indices = np.argsort(scores)[::-1][:top_k]

    results = semantic_df.iloc[best_indices].copy()
    results["similarity_score"] = scores[best_indices]

    return results[
        ["reference", "translation", "explanation",
         "similarity_score"]
    ].reset_index(drop=True)

# 5. RAMAYANA QUESTION ANSWERING ENGINE
def ask_ramayana(question, top_k=5):
    results = semantic_search(question, top_k=top_k)

    display(Markdown(
        f"## Ramayana Intelligence Response\n"
        f"**Question:** {question}"
    ))

    for index, row in results.iterrows():
        display(Markdown(
            f"""
### Evidence {index + 1}

**Reference:** {row['reference']}

**Translation:** {row['translation']}

**Explanation:** {row['explanation']}

**Semantic similarity:** {row['similarity_score']:.3f}
"""
        ))

    print(
        "\nNote: This system retrieves relevant verses. "
        "It does not independently generate or verify a definitive answer."
    )

    return results

# 6. CHARACTER COMPARISON ENGINE
def compare_characters(character_a, character_b):
    if character_a not in G:
        print(f"{character_a} is not present in the knowledge graph.")
        return

    if character_b not in G:
        print(f"{character_b} is not present in the knowledge graph.")
        return

    neighbors_a = set(G.neighbors(character_a))
    neighbors_b = set(G.neighbors(character_b))

    common = neighbors_a.intersection(neighbors_b)

    print("\nCHARACTER COMPARISON")
    print("=" * 50)
    print("Character A:", character_a)
    print("Character B:", character_b)
    print("Connections A:", len(neighbors_a))
    print("Connections B:", len(neighbors_b))
    print("Shared connections:", len(common))

    print("\nShared entities:")
    for entity in sorted(common):
        print("-", entity)

    try:
        path = nx.shortest_path(
            G.to_undirected(),
            source=character_a,
            target=character_b
        )
        print("\nShortest relationship path:")
        print(" → ".join(path))
    except nx.NetworkXNoPath:
        print("\nNo connecting path found.")

# 7. GRAPH INFLUENCE ANALYTICS
def advanced_graph_analytics():
    graph = G.to_undirected()

    pagerank = nx.pagerank(graph)
    betweenness = nx.betweenness_centrality(graph)
    degree = nx.degree_centrality(graph)
    closeness = nx.closeness_centrality(graph)

    analytics = pd.DataFrame({
        "entity": list(graph.nodes()),
        "pagerank": [pagerank[n] for n in graph.nodes()],
        "betweenness": [betweenness[n] for n in graph.nodes()],
        "degree_centrality": [degree[n] for n in graph.nodes()],
        "closeness": [closeness[n] for n in graph.nodes()]
    })

    analytics = analytics.sort_values(
        "pagerank", ascending=False
    ).reset_index(drop=True)

    return analytics

# 8. SAVE ADVANCED OUTPUTS
analytics_df = advanced_graph_analytics()

analytics_df.to_csv(
    "DharmaGraph_Advanced_Analytics.csv",
    index=False
)

semantic_df.to_csv(
    "DharmaGraph_Semantic_Index.csv",
    index=False
)

np.save(
    "DharmaGraph_Verse_Embeddings.npy",
    verse_embeddings
)

print("\nAdvanced analytics generated successfully!")

display(analytics_df.head(15))

print("\nAVAILABLE FUNCTIONS")
print("1. semantic_search('meaning of a query')")
print("2. ask_ramayana('your question')")
print("3. compare_characters('Rama', 'Hanuman')")
print("4. advanced_graph_analytics()")
```

```python

# ============================================================
# DHARMAGRAPH 2.0 — CHARACTER INTELLIGENCE ENGINE
# ============================================================

import networkx as nx
import pandas as pd
import numpy as np
import json
import os

os.makedirs("DharmaGraph/advanced_outputs", exist_ok=True)

# 1. ADVANCED GRAPH ANALYTICS

# Convert to a simple, undirected Graph (collapsing multi-edges for metrics like clustering)
analysis_graph = nx.Graph(G)

pagerank = nx.pagerank(analysis_graph)
betweenness = nx.betweenness_centrality(analysis_graph)
closeness = nx.closeness_centrality(analysis_graph)
degree = nx.degree_centrality(analysis_graph)
clustering = nx.clustering(analysis_graph)

graph_metrics = pd.DataFrame({
    "Entity": list(analysis_graph.nodes()),
    "PageRank": [pagerank[n] for n in analysis_graph.nodes()],
    "Betweenness": [betweenness[n] for n in analysis_graph.nodes()],
    "Closeness": [closeness[n] for n in analysis_graph.nodes()],
    "Degree_Centrality": [degree[n] for n in analysis_graph.nodes()],
    "Clustering_Coefficient": [
        clustering[n] for n in analysis_graph.nodes()
    ]
})

graph_metrics = graph_metrics.sort_values(
    "PageRank", ascending=False
).reset_index(drop=True)

print("TOP 15 MOST INFLUENTIAL GRAPH ENTITIES")
display(graph_metrics.head(15))


# 2. CHARACTER INTELLIGENCE PROFILE

def character_intelligence(character):

    if character not in G:
        print("Character not found:", character)
        return None

    neighbors = list(G.neighbors(character))

    profile = {
        "Character": character,
        "Total_Connections": len(neighbors),
        "PageRank": pagerank.get(character, 0),
        "Betweenness": betweenness.get(character, 0),
        "Degree_Centrality": degree.get(character, 0),
        "Connected_Entities": neighbors
    }

    print("\nCHARACTER INTELLIGENCE PROFILE")
    print("=" * 45)

    for key, value in profile.items():
        print(f"{key}: {value}")

    return profile


# 3. RELATIONSHIP PATH FINDER

def relationship_path(character_a, character_b):

    if character_a not in analysis_graph:
        print(character_a, "not found.")
        return

    if character_b not in analysis_graph:
        print(character_b, "not found.")
        return

    try:
        path = nx.shortest_path(
            analysis_graph,
            source=character_a,
            target=character_b
        )

        print("\nRELATIONSHIP PATH")
        print("=" * 45)
        print(" → ".join(path))
        print("Path length:", len(path) - 1)

        return path

    except nx.NetworkXNoPath:
        print("No relationship path exists.")
        return None


# 4. COMMON CONNECTION ANALYSIS

def common_connections(character_a, character_b):

    if character_a not in analysis_graph or character_b not in analysis_graph:
        print("One or both entities are missing.")
        return

    a = set(analysis_graph.neighbors(character_a))
    b = set(analysis_graph.neighbors(character_b))

    common = a.intersection(b)

    print("\nCOMMON CONNECTION ANALYSIS")
    print("=" * 45)
    print("Character A:", character_a)
    print("Character B:", character_b)
    print("Shared connections:", len(common))

    for entity in sorted(common):
        print("-", entity)

    return sorted(common)


# 5. GRAPH COMMUNITY DETECTION

communities = list(
    nx.community.greedy_modularity_communities(
        analysis_graph
    )
)

community_rows = []

for community_number, community in enumerate(communities, start=1):

    for entity in community:
        community_rows.append({
            "Community": community_number,
            "Entity": entity
        })

communities_df = pd.DataFrame(community_rows)

print("\nDETECTED GRAPH COMMUNITIES")
display(communities_df.head(30))


# 6. SAVE ALL ADVANCED RESULTS

graph_metrics.to_csv(
    "DharmaGraph/advanced_outputs/graph_intelligence_metrics.csv",
    index=False
)

communities_df.to_csv(
    "DharmaGraph/advanced_outputs/graph_communities.csv",
    index=False
)

# Export character profiles
character_profiles = {}

for node, attributes in G.nodes(data=True):

    if attributes.get("type") == "Character":

        character_profiles[node] = {
            "connections": list(G.neighbors(node)),
            "pagerank": pagerank.get(node, 0),
            "betweenness": betweenness.get(node, 0),
            "degree_centrality": degree.get(node, 0)
        }

with open(
    "DharmaGraph/advanced_outputs/character_profiles.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        character_profiles,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\nADVANCED CHARACTER INTELLIGENCE COMPLETED!")
print("Generated:")
print("- graph_intelligence_metrics.csv")
print("- graph_communities.csv")
print("- character_profiles.json")
```
